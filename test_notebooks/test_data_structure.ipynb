```python
from astropy.io import fits
import os
from glob import glob
import pandas as pd
from os.path import basename


master_dir = '/data/beegfs/astro-storage/groups/beuther/pasteimle/HD1461_grouped/'

def make_flat_DF():
    
    star = 'FLAT'
    star_parent_dir = master_dir + star + '/IRDIS/'
    all_obs_nights = os.listdir(star_parent_dir)

    FLAT_DF = pd.DataFrame()

    for obs_night in all_obs_nights:

        obs_night_parent_dir = star_parent_dir + obs_night + '/'
        
        filters = os.listdir(obs_night_parent_dir)

        for filter_name in filters:

            filter_parent_dir = obs_night_parent_dir + filter_name + '/'

            all_files = sorted(glob(filter_parent_dir + '*.fits'))

            for file in all_files:
                print(file)
                filename = basename(file) 
                exp_time = fits.getheader(file)['EXPTIME']
                
                DF_i = pd.DataFrame({'filename': [filename], 
                                    'obs_night': [obs_night],
                                    'obs_run': [obs_night[:-3]],
                                    'filter': [filter_name],
                                    'exp_time': [exp_time], 
                                    'filepath': [file],

                                    })

                FLAT_DF = pd.concat((FLAT_DF, DF_i), ignore_index=True)

    return FLAT_DF

def make_dark_DF():
    DARK_DF = pd.DataFrame()
    star = 'DARK'
    star_parent_dir = master_dir + star + '/IRDIS/'
    all_obs_nights = os.listdir(star_parent_dir)

    for obs_night in all_obs_nights:

        obs_night_parent_dir = star_parent_dir + obs_night + '/'
        
        filters = os.listdir(obs_night_parent_dir)

        for filter_name in filters:

            filter_parent_dir = obs_night_parent_dir + filter_name + '/'

            all_files = sorted(glob(filter_parent_dir + '*.fits'))

            for file in all_files:
                print(file)
                
                filename = basename(file) 
                hdul = fits.open(file)

                
                #exp_time = fits.getheader(file)['EXPTIME']
                
                # DF_i = pd.DataFrame({'filename': [filename], 
                #                     'obs_night': [obs_night],
                #                     'obs_run': [obs_night[:-3]],
                #                     'filter': [filter_name],
                #                     'exp_time': [exp_time], 
                #                     'filepath': [file],

                #                     })

                # DARK_DF = pd.concat((DARK_DF, DF_i), ignore_index=True)
                
    return DARK_DF




def star_center_dict():

    DF = pd.DataFrame({})
    all_stars = [f for f in os.listdir(master_dir) if f not in ['FLAT','DARK', 'DB_H23','DISTORT']]
    for star in all_stars:      
        print(star)
        star_parent_dir = master_dir + star + '/IRDIS/'
        all_obs_nights = os.listdir(star_parent_dir)

        FLAT_DF = pd.DataFrame()

        for obs_night in all_obs_nights:

            obs_night_parent_dir = star_parent_dir + obs_night + '/'
            
            filters = os.listdir(obs_night_parent_dir)

            for filter_name in filters:

                filter_parent_dir = obs_night_parent_dir + filter_name + '/'

                all_files = sorted(glob(filter_parent_dir + '*.fits'))
                star_center_files = [f for f in all_files if 'STAR_CENTER' in f]

                if len (star_center_files) == 0:

                    
                    DF_i = pd.DataFrame({'star': [star],
                                    'obs_night': [obs_night],
                                    'obs_run': [obs_night[:-3]],
                                    'filter': [filter_name]
                                    })
                    
                    DF = pd.concat((DF, DF_i), ignore_index=True)

    return DF

DF = star_center_dict()
DF.to_csv('/home/pasteimle/code/starcenter.csv')
```
