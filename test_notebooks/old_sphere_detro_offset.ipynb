```python
from astropy.io import fits 
from glob import glob 
import numpy as np
import os
```

```python
folder_path = '/nexus/posix0/MIA-astro-env/lkreidberg/pasteimle/additional_ISPY_raw_files/HD179218/IRDIS/2016-05-21/DB_H23'
all_files = sorted(glob(folder_path + '/SC*.fits'))
```

```python
all_corrections = []
for file in all_files:
    header = fits.getheader(file)

    correction = np.arctan(np.tan((header['HIERARCH ESO TEL ALT'] - 2 * header['HIERARCH ESO INS4 DROT2 BEGIN']) * np.pi / 180 ) ) * 180 / np.pi
    print(os.path.basename(file), correction)
    all_corrections.append(correction)

print(np.mean(all_corrections))
```

```python
percentage = np.mean(all_corrections) / 360

total_rad = 512*2*np.pi

pixel_offset = percentage * total_rad
pixel_offset
```
